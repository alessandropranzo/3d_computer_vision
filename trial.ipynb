```python
import os
import torch
import open3d as o3d
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
from tqdm import tqdm
from vggt.models.vggt import VGGT
from vggt.utils.load_fn import load_and_preprocess_images
from vggt.utils.visual_track import visualize_tracks_on_images
```

```python
def plot_3d_representation(dir_path, subsample_thr=100_000, plot_traj=True):
    full_pcd_path = os.path.join(dir_path, "scene_representation.ply")
    full_traj_path = os.path.join(dir_path, "camera_trajectory.npy")

    pcd = o3d.io.read_point_cloud(full_pcd_path)
    print(pcd)

    traj = np.load(full_traj_path)

    points_3d = np.asarray(pcd.points)
    colors = np.asarray(pcd.colors)

    if len(points_3d) > subsample_thr:
        indices = np.random.choice(len(points_3d), subsample_thr, replace=False)
        points_3d = points_3d[indices]
        if colors is not None:
            colors = colors[indices]
        print(f"Subsampling the plot to {subsample_thr} points")

    data_list = [go.Scatter3d(
        x=points_3d[:, 0], 
        y=points_3d[:, 1], 
        z=points_3d[:, 2],
        mode='markers',
        marker=dict(
            size=1.5,
            color=colors,
            opacity=1.0
        )
    )]

    if plot_traj:
        data_list.append(
            go.Scatter3d(
                x=traj[:, 0],
                y=traj[:, 1],
                z=traj[:, 2],
                mode='lines+markers',
                marker=dict(size=4, color='red'),
                line=dict(color='red', width=2),
                name='Camera Trajectory'
            )
        )

    fig = go.Figure(data=data_list)

    fig.update_layout(scene=dict(aspectmode='data'), margin=dict(l=0, r=0, b=0, t=0))
    fig.show()
```

```python
plot_3d_representation("./predictions_visuals/room1")
```
