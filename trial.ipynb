```python
import open3d as o3d

# Load the file
pcd = o3d.io.read_point_cloud("/users/eleves-b/2024/mattia.martino/3d_computer_vision/predictions_visuals/scene_representation.ply")

# Print some info
print(pcd)
```

```python
traj_path = "/users/eleves-b/2024/mattia.martino/3d_computer_vision/predictions_visuals/camera_trajectory.npy"
if os.path.exists(traj_path):
    trajectory = np.load(traj_path)
else:
    trajectory = None
```

```python
import numpy as np
import plotly.graph_objects as go


points_3d = np.asarray(pcd.points)
colors = np.asarray(pcd.colors)

threshold = 500_000

if len(points_3d) > threshold:
    indices = np.random.choice(len(points_3d), threshold, replace=False)
    points_3d = points_3d[indices]
    if colors is not None:
        colors = colors[indices]

# 5. Plot
data_list = [go.Scatter3d(
    x=points_3d[:, 0], 
    y=points_3d[:, 1], 
    z=points_3d[:, 2],
    mode='markers',
    marker=dict(
        size=1.5,
        color=colors,
        opacity=1.0
    )
)]

if trajectory is not None:
    data_list.append(
        go.Scatter3d(
            x=trajectory[:, 0],
            y=trajectory[:, 1],
            z=trajectory[:, 2],
            mode='lines+markers',
            marker=dict(size=4, color='red'),
            line=dict(color='red', width=2),
            name='Camera Trajectory'
        )
    )

fig = go.Figure(data=data_list)

fig.update_layout(scene=dict(aspectmode='data'), margin=dict(l=0, r=0, b=0, t=0))
fig.show()
```
